# Live lab presets

Each `launch_lab(...)` cell **opens Pygame and blocks** until you close the window (Esc).

Run **one** launch cell at a time. Edit the kwargs if you want a different density, seed, or agent count.

```python
from pathlib import Path
import os

if Path.cwd().name == "notebooks":
    os.chdir(Path.cwd().parent)

from swarmpathfinding.viz.live import launch_lab, launch_spec
from swarmpathfinding.viz.settings import PRESETS, LabSettings
from swarmpathfinding.generators.packs import pack_specs

print("presets:", ", ".join(PRESETS))
```

## Fog of war — shared last-seen, clump of scouts

Walk (unpause) and watch the fog peel back. A wall you have not seen yet is assumed walkable (freespace).

```python
launch_lab(PRESETS["fog_scout"])
```

## RTS flanks — street map, 30 units, two spawn sides, same goal

```python
launch_lab(PRESETS["rts_flanks"])
```

## Opposite corridor — two groups, two goals

```python
launch_lab(PRESETS["opposite_corridor"])
```

## Slow edit — rooms, one agent. Press **E** to delete the pending wall.

```python
launch_lab(PRESETS["slow_edit_rooms"])
```

## Continuous U-trap (potential-field killer) and moving obstacles behind fog

```python
launch_lab(PRESETS["utrap"])
```

```python
launch_lab(PRESETS["movers_ghost"])
```

## Custom settings

Any field on `LabSettings` can be overridden.

```python
launch_lab(
    LabSettings(
        world_kind="grid",
        map_kind="game",
        planner="straight_line",
        starts="scattered",
        n_agents=12,
        density=0.15,
        fow=True,
        vision="cone",
        seed=7,
    )
)
```

## Launch a pack row

Pick a pack, pick an index, open it in Pygame.

```python
PACK = "fow_discover"
INDEX = 0

specs = pack_specs(PACK, seeds=(0,))
for i, spec in enumerate(specs):
    mark = "<--" if i == INDEX else ""
    print(f"{i:3d} {spec.scenario_id} {mark}")
```

```python
launch_spec(specs[INDEX])
```
