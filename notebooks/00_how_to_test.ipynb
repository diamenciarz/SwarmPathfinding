No window. Uses stub planners until A* / D* Lite exist.


Open [02_run_pack.ipynb](02_run_pack.ipynb) to pick a pack and see a table.

````python
from swarmpathfinding.generators.packs import list_packs, pack_specs

for name in list_packs():
    n = len(pack_specs(name, seeds=(0,)))
    print(f"{name:18s} {n:3d} scenarios (seed 0)")
````

## Live window keys

| Key | Action |
|---|---|
| LMB / RMB | add / delete wall |
| G+LMB | set goal |
| E | apply pending slow-edit (delete the pre-chosen wall) |
| Space | pause |
| . | step |
| R | regenerate |
| Esc | quit |
